1. Считайте датасет из файла train.csv (это данные о выживаемости на Титанике) с помощью pandas

```python
import pandas as pd
import bottleneck as bn

df = pd.read_csv('train.csv')
df.head()
```

2. Посчитайте средний возраст пассажиров и его стандартное отклонение с помощью bottleneck

```python
mean_age = bn.nanmean(df["Age"])
std_age = bn.nanstd(df["Age"])

print("Средний возраст пассажиров:", mean_age, sep=' ')
print("Стандартное отклонение возраста:", std_age, sep=' ')
```

3. Для каждого пассажира умножьте значение столбца Fare на 1.3 и сохраните результаты как новый столбец Fare_new

```python
fare_new = []
for row in df.itertuples():
    fare_new.append(row.Fare * 1.3)
df["Fare_new"] = fare_new

print("Таблица с новым столбцом Fare_new (через itertuples):")
df[["PassengerId", "Fare", "Fare_new"]].head()
```

2. Для каждого столбца определите оптимальный с точки зрения потребления памяти тип данных - напишите свои выводы в комментариях

```python
df.info(memory_usage="deep")

# Анализ каждого столбца
for column in df.columns:
    print(f"\nСтолбец: {column}")
    print(f"Текущий тип: {df[column].dtype}")
    print(f"Уникальных значений: {df[column].nunique()}")
    print(f"Пример значений: {df[column].iloc[:5].tolist()}")
```

Анализ и выбор типов:
1. price (int64): Цены домов от 1750000 до 13300000. Используем int32 (диапазон -2^31 до 2^31-1, т.е. около -2.1B до 2.1B), так как максимальное значение 13300000 укладывается.
2. area (int64): Площадь от 1650 до 16200. Используем int32, так как максимум 16200 далеко от предела int32.
3. bedrooms (int64): Количество спален от 1 до 6. Используем int8 (диапазон -128 до 127), так как максимум 6.
4. bathrooms (int64): Ванные от 1 до 4. Используем int8, так как максимум 4.
5. stories (int64): Этажи от 1 до 4. Используем int8, так как максимум 4.
6. mainroad (object): Два значения "yes"/"no". Преобразуем в category (экономия памяти для категориальных данных).
7. guestroom (object): "yes"/"no". Преобразуем в category.
8. basement (object): "yes"/"no". Преобразуем в category.
9. hotwaterheating (object): "yes"/"no". Преобразуем в category.
10. airconditioning (object): "yes"/"no". Преобразуем в category.
11. parking (int64): Парковочные места от 0 до 3. Используем int8, так как максимум 3.
12. prefarea (object): "yes"/"no". Преобразуем в category.
13. furnishingstatus (object): Три значения "furnished"/"semi-furnished"/"unfurnished". Преобразуем в category.


3. Поменяйте типы данных столбцов датафрейма на выбранные вами в прошлом пункте и сравните потребление памяти до и после оптимизации

```python
memory_before = df.memory_usage(deep=True).sum() / 1024  # в KB

# Оптимизация типов данных
df_optimized = df.astype({
    "price": "int32",
    "area": "int32",
    "bedrooms": "int8",
    "bathrooms": "int8",
    "stories": "int8",
    "mainroad": "category",
    "guestroom": "category",
    "basement": "category",
    "hotwaterheating": "category",
    "airconditioning": "category",
    "parking": "int8",
    "prefarea": "category",
    "furnishingstatus": "category"
})

# Потребление памяти после оптимизации
memory_after = df_optimized.memory_usage(deep=True).sum() / 1024  # в KB

# Вывод результатов
display("Потребление памяти до оптимизации (KB):", memory_before)
display("Потребление памяти после оптимизации (KB):", memory_after)
display("Экономия памяти (%):", (memory_before - memory_after) / memory_before * 100)

# Проверка структуры оптимизированного датафрейма
display("Информация о датасете после оптимизации:")
display(df_optimized.info(memory_usage="deep"))
```
