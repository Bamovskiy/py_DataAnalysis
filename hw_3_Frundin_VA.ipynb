3. Посчитайте процент выживаемости у каждого класса пассажиров (Pclass)

```python
Survived_by_class = Data.groupby('Pclass')['Survived'].sum()
Survived_by_class
```

```python
All_by_class = Data['Pclass'].value_counts()
All_by_class
```

```python
Survived_in_percent = (Survived_by_class / All_by_class) * 100
Survived_in_percent
```

4. Выведите самое популярное мужское и самое популярное женское имя на корабле

```python
def extract_first_name(full_name):
    parts = full_name.split(', ')
    if len(parts) > 1:
        name_part = parts[1]
        name = name_part.split(' ')[1] # убираем Mr, Mrs, Miss
        return name
    return None

Data['First_Name'] = Data['Name'].apply(extract_first_name)
Data[Data['Sex'] == 'male']['First_Name'].value_counts().head(1)
```

```python
Data[Data['Sex'] == 'female']['First_Name'].value_counts().head(1)
```
